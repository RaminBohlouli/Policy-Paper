```python
import pandas as pd
import os
import json

# Set the current working directory
current_dir = os.getcwd()

# File path for the input CSV
file_path = os.path.join(current_dir, r"organized\2_2022_translated.csv")

# Load the CSV file into a DataFrame
df = pd.read_csv(file_path, encoding="ISO-8859-1")

# Load the column mapping with metadata from the JSON file
mapping_path = os.path.join(current_dir, r"Helper Data\column_mappings_new.json")
with open(mapping_path, "r", encoding="ISO-8859-1") as f:
    mapping_data = json.load(f)

# Create a dictionary for renaming columns
rename_dict = {
    old_name: meta["new_name"]
    for old_name, meta in mapping_data.items()
}

# Columns that exist in both the CSV and the JSON mapping
mapped_cols = [col for col in df.columns if col in rename_dict]

# Columns that exist in the CSV but NOT in the JSON mapping
unmapped_cols = [col for col in df.columns if col not in rename_dict]

# Reorder columns: mapped first, unmapped at the end
df = df[mapped_cols + unmapped_cols]

# Rename only the mapped columns
df.rename(columns=rename_dict, inplace=True)

# Save the cleaned DataFrame to a new CSV file
output_path = os.path.join(current_dir, r"organized\3_2022_columnNamesMapped.csv")
df.to_csv(output_path, encoding="ISO-8859-1", index=False)
```
