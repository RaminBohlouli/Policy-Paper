```python
import pandas as pd

# Input and output file paths
input_csv = "2022.csv"
output_csv = r"organized\1_2022_organized.csv"

df = pd.read_csv(input_csv, encoding='ISO-8859-1')

# Clean column names
df.columns = df.columns.str.strip()

# Replace empty clonedfrom with NaN
df["clonedfrom"] = df["clonedfrom"].replace(r"^\s*$", pd.NA, regex=True)

# Determine the final question id
df["final_question_id"] = df["clonedfrom"].fillna(df["question"])

# Columns that must remain per employee
index_columns = [
    "anonid",
    "questionnaire",
    "office_id",
    "office_name",
    "lon_office",
    "lat_office",
    "lon_home",
    "lat_home"
]

# Pivot
pivoted = df.pivot_table(
    index=index_columns,
    columns="final_question_id",
    values="val",
    aggfunc=lambda x: " | ".join(x.astype(str).dropna().unique())
).reset_index()

# Remove column group name
pivoted.columns.name = None

# Save result
pivoted.to_csv(output_csv, index=False, encoding="ISO-8859-1")

print("Pivot completed.")
print(pivoted.head())
```
