```python
import pandas as pd

# File paths
file_2023 = r"C:\Users\Ramin\source\repos\Research Repo\ModeChoiceHybrid\Fastweb\Data\organized\5_2023_urbanCategoryFound.csv"
file_2024 = r"C:\Users\Ramin\source\repos\Research Repo\ModeChoiceHybrid\Fastweb\Data\organized\5_2024_urbanCategoryFound.csv"
output_file = "employees_changed_mode.csv"

# Read CSVs with the specified encoding
df_2023 = pd.read_csv(file_2023, encoding="ISO-8859-1")
df_2024 = pd.read_csv(file_2024, encoding="ISO-8859-1")

# Keep only the needed columns
cols = ["employee", "office_location_type", "home_location_type", "main_mode"]
df_2023 = df_2023[cols].copy()
df_2024 = df_2024[cols].copy()

# Rename columns so they are distinguishable after merge
df_2023 = df_2023.rename(columns={
    "office_location_type": "2023_office_location_type",
    "home_location_type": "2023_home_location_type",
    "main_mode": "2023_mode"
})

df_2024 = df_2024.rename(columns={
    "office_location_type": "2024_office_location_type",
    "home_location_type": "2024_home_location_type",
    "main_mode": "2024_mode"
})

# Merge only employees existing in both files
merged = pd.merge(df_2023, df_2024, on="employee", how="inner")

# Keep only employees whose mode changed
changed_mode = merged[merged["2023_mode"] != merged["2024_mode"]].copy()

# Optional: reorder columns
changed_mode = changed_mode[
    [
        "employee",
        "2023_office_location_type",
        "2024_office_location_type",
        "2023_home_location_type",
        "2024_home_location_type",
        "2023_mode",
        "2024_mode"
    ]
]

# Save output CSV
changed_mode.to_csv(output_file, index=False, encoding="ISO-8859-1")

print(f"Done. Output saved as: {output_file}")
print(f"Employees with changed mode: {len(changed_mode)}")
```
